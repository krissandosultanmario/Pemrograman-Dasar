```python
# Latihan 1:
tegangan_panel = 18             # Nilai tegangan (Volt)
arus_panel = 6                  # Nilai arus (Ampere)
jenis_sel = "Monocrystalline"   # Spesifikasi tipe silikon sel surya

print("Tegangan:", tegangan_panel, "V")
print("Arus:", arus_panel, "A")
print("Jenis Sel:", jenis_sel)
```

```python
# Latihan 2:
nama_komponen = "kapasitor" #str
nilai_tegangan = 50  #int
nilai_resistansi = 10.5 #float
kondisi_baik = True  #bool

print("nama komponen =",nama_komponen,type(nama_komponen))
print("nilai tegangan =",nilai_tegangan,type(nilai_tegangan))
print("nilai resistansi =",nilai_resistansi,type(nilai_resistansi))
print("kondisi baik =",kondisi_baik,type(kondisi_baik))
```

```python
# Latihan 3:
nama_unit = "PLTS Nunukan"      # String: identitas teks unit PLTS
jumlah_panel = 240              # Integer: kuantitas fisik modul panel
tegangan_keluaran = 48.5        # Float: hasil ukur voltmeter (Volt)
sedang_beroperasi = True        # Boolean: status biner beroperasi (True/False)

print("nama_unit         :", nama_unit, "->", type(nama_unit))
print("jumlah_panel      :", jumlah_panel, "->", type(jumlah_panel))
print("tegangan_keluaran :", tegangan_keluaran, "->", type(tegangan_keluaran))
print("sedang_beroperasi :", sedang_beroperasi, "->", type(sedang_beroperasi))


```

```python
# Latihan 4:
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Mengakses indeks awal dan akhir
print("Pembacaan pertama :", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[-1])

# (b) Menyisipkan data baru di urutan paling belakang
tegangan_harian.append(222.0)
print("List setelah ditambah:", tegangan_harian)

# (c) Mengukur panjang list
print("Jumlah total data    :", len(tegangan_harian))
```

```python
# Latihan 5:
motor_listrik = {
    "nama": "Motor Induksi 3 Fasa",
    "daya_kW": 15.0,
    "tegangan_V": 380,
    "rpm": 1450
}

# Mengakses data spesifik melalui kunci
print("RPM awal:", motor_listrik["rpm"])

# Memperbarui nilai RPM secara dinamis
motor_listrik["rpm"] = 1500

# Menampilkan struktur dictionary yang telah termodifikasi
print("Data motor setelah diubah:")
print(motor_listrik)
```

```python
# Latihan 6:
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

# 1. Akses nama unit kedua (indeks 1, key 'nama')
print("Nama unit kedua:", pembangkit[1]["nama"])

# 2. Mengubah status unit ketiga menjadi operasi
pembangkit[2]["status"] = "operasi"

# 3. Menambahkan unit baru ke koleksi pembangkit
unit_baru = {"nama": "Unit-4", "kapasitas_MW": 200, "status": "standby"}
pembangkit.append(unit_baru)

# 4. Mencetak keseluruhan data pembangkit terkini
print("\nSeluruh data pembangkit:")
print(pembangkit)
```
