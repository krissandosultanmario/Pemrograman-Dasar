```python
# Latihan

tahun = int(input("Masukkan tahun: "))

if (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0):
    print(f"{tahun} adalah tahun kabisat.")
else:
    print(f"{tahun} bukan tahun kabisat.")
```

```python
# Latihan

panjang = float(input("Masukkan panjang: "))
lebar = float(input("Masukkan lebar: "))

luas = panjang * lebar
keliling = 2 * (panjang + lebar)

print(f"Luas persegi panjang: {luas}")
print(f"Keliling persegi panjang: {keliling}")
```

```python
# Soal 1 : Belanja Buku

jumlah_buku = int(input("Jumlah buku: "))
harga_buku = int(input("Harga per buku: "))
uang_budi = int(input("Uang bayar: "))

total = jumlah_buku * harga_buku
diskon = total * 10 // 100
bayar = total - diskon
kembalian = uang_budi - bayar

per_orang = bayar // 4
sisa = bayar % 4
uang_cukup = uang_budi >= bayar

print(f"Total       : {total}")
print(f"Diskon      : {diskon}")
print(f"Bayar       : {bayar}")
print(f"Kembalian   : {kembalian}")
print(f"Per orang   : {per_orang}")
print(f"Sisa        : {sisa}")
print(f"Uang cukup? : {uang_cukup}")
```

```python
# SOAL 2 : PLTS Atap Rumah

jumlah_panel = int(input("Jumlah panel: "))
daya_panel = float(input("Daya per panel (Wp): "))
jam_sinar = float(input("Penyinaran efektif (jam): "))
efisiensi = float(input("Efisiensi sistem: "))
kebutuhan = float(input("Kebutuhan listrik (kWh): "))
tarif = float(input("Tarif listrik per kWh: "))

energi_harian = round((jumlah_panel * daya_panel * jam_sinar * efisiensi) / 1000, 2)
energi_bulanan = round(energi_harian * 30, 2)
nilai_listrik = round(energi_bulanan * tarif, 2)
kekurangan = round(kebutuhan - energi_harian, 2)
mencukupi = energi_harian >= kebutuhan

print(f"Energi harian (kWh) : {energi_harian}")
print(f"Energi bulanan (kWh): {energi_bulanan}")
print(f"Nilai listrik (Rp)  : {nilai_listrik}")
print(f"Kekurangan (kWh)    : {kekurangan}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
# Soal 3 : Generator Diesel dan Rugi Kabel

daya = int(input("Daya generator (kW): "))
jam = int(input("Jam operasi per hari: "))
konsumsi = float(input("Konsumsi solar per kWh: "))
harga_solar = float(input("Harga solar per liter: "))
kapasitas_tangki = float(input("Kapasitas tangki (liter): "))
arus = float(input("Arus kabel (A): "))
hambatan = float(input("Hambatan kabel (ohm): "))

energi = daya * jam
solar = energi * konsumsi
biaya = solar * harga_solar
tangki_cukup = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar
rugi_kabel = (arus ** 2) * hambatan
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print(f"Energi (kWh)    : {energi}")
print(f"Solar (liter)   : {solar}")
print(f"Biaya (Rp)      : {biaya}")
print(f"Tangki cukup    : {tangki_cukup} hari")
print(f"Sisa solar (L)  : {sisa_solar}")
print(f"Rugi kabel (W)  : {rugi_kabel}")
print(f"Rugi tinggi?    : {rugi_tinggi}")
```
