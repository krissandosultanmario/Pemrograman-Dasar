```python
def is_leap_year(year):
    """
    Menentukan apakah suatu tahun merupakan tahun kabisat.
    Aturan:
    - Habis dibagi 4 dan tidak habis dibagi 100, atau
    - Habis dibagi 400
    """
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

# Program utama
try:
    tahun = int(input("Masukkan tahun (4 digit): "))
    if tahun < 0:
        print("Tahun tidak boleh negatif.")
    else:
        if is_leap_year(tahun):
            print(f"Tahun {tahun} adalah tahun kabisat.")
        else:
            print(f"Tahun {tahun} bukan tahun kabisat.")
except ValueError:
    print("Input harus berupa angka bulat.")
```
